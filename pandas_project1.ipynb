## MYNTRA SALES ANALYSIS

This project analyzes Myntra sales data to uncover insights about product categories, city-wise sales, and overall trends.  
We aim to answer questions like:  
- Which category sells the most?  
- Which city contributes the highest sales?  
- How do discounts and quantities impact revenue?  
- What are the sales trends over time?  

The analysis is done using **Python (Pandas).



```python
import pandas as pd
```

## <u> Category wise Analysis </u>

category wise Analysis helps us to understand which category sales is highest and which category sales is lowest . So that we work on it .

```python
category_insights = df.groupby("Category")["Final_Amount (₹)"].sum().sort_values(ascending=False)
print(category_insights)
Best_selling_category = category_insights.idxmax()
Total_Highestsales_= category_insights.max() 
print("Best category is : ",Best_selling_category)
print("Total sales of best category is :",Total_Highestsales_)
```

```python
import pandas as pd 
df = pd.read_csv(r"Data\myntra_sales.csv")
import matplotlib.pyplot as plt
Brand = df.groupby("Category")["Final_Amount (₹)"].sum()
plt.figure(figsize=(8,6))
plt.bar(Brand.index,Brand.values,label="category-sales",color='green')
plt.xlabel("category-type",fontsize=25,color='blue')
plt.ylabel("Total sales",fontsize=25,color='blue')
plt.title("CATEGORY WISE ANALYSIS",fontweight='bold',fontsize= 35)
plt.grid(color='lightgreen',linestyle=':')
plt.legend()
plt.tight_layout()
plt.savefig('category.png')
plt.show()
```

## <u> Location Wise Analysis Of Sales.</u>

Location wise Analysis sales help us to find which city have highest sales .


```python
location_insight = df.groupby("City")["Final_Amount (₹)"].sum().sort_values(ascending=False)
print(location_insight)
Best_selling_city= location_insight.idxmax()
Worst_selling_city = location_insight.idxmin()
Total_bestcity_sales= location_insight.max()
Total_worstcity_sales= location_insight.min()
print("Best selling city is :",Best_selling_city)
print("Total sales of best selling city is : ",Total_bestcity_sales)
print("Worst selling city is : ", Worst_selling_city)
print("Total sales of worst selling city is :",Total_worstcity_sales)
```

## <u> Weekly Sales Trend </u>

To understand how sales change over time, we analyze the data on a **weekly basis**.  

Steps:  
1. Convert the `Order_Date` column into a proper **datetime format**.  
2. Group the sales by **weekly** using Pandas `.dt.to_period("W")`.   

This analysis helps identify:  
- Seasonal demand patterns 📈  
- High-performing and low-performing Weekly.
- Insights for planning marketing campaigns and inventory management



```python
# Here first we convert Date datatype into Datetime datatype.
df["Date"] = pd.to_datetime(df["Date"])
```

```python
# Now we check our Date datatype change or not 
print(df.info())
```

```python
sales_trend =df.groupby(df["Date"].dt.to_period("W"))["Final_Amount (₹)"].sum().sort_values(ascending=False)
print(sales_trend)
max_sales = sales_trend.idxmax()
min_sales = sales_trend.idxmin()
maxnumber_of_sales = sales_trend.max()
minnumber_of_sales = sales_trend.min()
print(" Weekly Maximum sales on :",max_sales, "is ",maxnumber_of_sales)
print("weakly minimum sales on  : ",min_sales, "is" ,minnumber_of_sales)
```

```python
import pandas as pd 
df = pd.read_csv(r'Data/myntra_sales.csv')
import matplotlib.pyplot as plt
df["Date"] = pd.to_datetime(df["Date"])
sales_trend =df.groupby(df["Date"].dt.to_period("W"))["Final_Amount (₹)"].sum()
plt.Figure(figsize=(8,6))
plt.barh(sales_trend.index.astype(str),sales_trend.values,color='pink')
plt.ylabel('weeks',fontsize=25,color='red')
plt.xlabel('sales',fontsize=25,color='red')
plt.title('WEEKLY-SALES ANALYSIS',fontsize=35,color='black')
plt.grid(linestyle=':',color='red')
plt.tight_layout()
plt.savefig('weekly-sales.png')
plt.show()

```
